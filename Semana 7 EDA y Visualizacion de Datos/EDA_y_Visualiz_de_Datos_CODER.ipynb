## 5. ¿Qué es una Variable?

En programación, se le asigna un nombre o "apodo" a cada pequeño segmento de la memoria que almacena un dato o un valor.
Una variable es un espacio reservado en memoria al que le ponemos un "apodo", en el que se puede almacenar algo (un elemento o dato) y que permite recuperar/transformar/modificar lo almacenado si lo volvemos a declarar una vez guardado previamente, invocándolo por su "apodo". 😵‍💫


Para que un programa funcione y pueda realizar tareas es necesario guardar en la memoria de la computadora los datos que necesita. De esta forma, cada vez que se usa este nombre en el programa se está refiriendo al valor del dato almacenado en la memoria. Este es el concepto de **"variable"**.

En python esto se logra de la siguiente manera:


```python
x = 'Esto es una "variable!"'
print(x)
```

Entonces, cuando al comando **`print()`** le "damos de comer" la variable **`x`**, lo que hace el programa es mostrarnos el contenido de esa variable (es decir, el dato que está almacenado en ese espacio en memoria) y no la letra "x".

**Cuidado!!** 🙆‍♂️

En estas variables podemos guardar lo que sea necesario para ejecutar el programa.
Notemos que para asignar una variable el símbolo que utilizamos fue el **`=`**. A este operador lo llamamos **operador de asignación**.

```python
x = '¡Hola Mundo!'
y = "¡Adiós Mundo cruel!"
z = 32765
print(x)
print(y)
print(z)
```

## 8. Operaciones Básicas con Variables


### 8.1. Operaciones de Modificación


Las operaciones básicas que pueden llevarse a cabo utilizando variables:

- Sumar dos variables ( $a+b$ ):

  `a + b`

- Restar dos variables ( $a-b$ ):

  `a − b`

- Multiplicar dos variables ( $a\cdot b$ ):

  `a * b`

- Dividir una variable por la otra ( $a\div b$ ):

  `a / b`

- Una variable elevada a la otra ( $a^b$ ):

  `a ** b`

- Asignarle un valor a una variable ( $a\leftarrow b$ ):

  `a = b`
  

¿Cual será la salida del siguiente programa?

```python
x = 3
y = -2
y = x*y
x = x**2
```

```python
print("y =", y)
print("x =", x)
```

```python
# no sólo sirve para números!
w = 'a'
w = 3*w
```

```python
print("w = ", w )
```

### 8.2. Operaciones de Comparación


Existen operaciones que permiten comparar propiedades de las variables. Las principales formas de comparación son:

*   Menor que ( $a < b$ ):

 `a < b`

*   Mayor que ( $a > b$ ):

 `a > b`

*   Menor o igual que ( $a \leq b$ ):

 `a <= b`

*   Mayor o igual que ( $a \geq b$ ):

 `a >= b`

*   Es igual ( $a = b$ ):

 `a == b` (Ojo: se utilizan 2 símbolos `=` para diferenciarse del operador asignación)

*   Es distinto ( $a \neq b$ ):

 `a != b`


Al comprarar dos números, se obtiene **`True`** o **`False`** (Verdadero o Falso) según si se cumple o no la condición establecida:

```python
x = 4
print("Es x distinto de 4?", x != 4)
print("Es x menor a 3?", x < 3)
```

También se pueden combinar diversas condiciones:

*  Si se cumplen las dos condiciones ( $a \, y \, b$ ):

  `a and b`

*  Si se cumple alguna de las condiciones ( $ a \, o \, b$ ):

  `a or b`

*  Si no se cumple esa condición ( no $ a $ ):

  `not a`


Veamos un ejemplo:

```python
x = int(input("Ingrese un número:"))
y = x > 5 and x < 7
z = x > 5 or x < 7
k = not x > 5
print("x > 5 y x < 7?", y)
print("x > 5 o x < 7?", z)
print("Es x NO mayor a 5?", k)
```

No sólo sirve para números, como vimos anteriormente, el lenguaje distingue entre mayúsculas y minúsculas (case sensitive):

```python
x = 'Hola'
y = 'hola'
x == y
```

```python
x = 'chau'
y = 'chau'
x == y
```

Estas condiciones pueden agruparse a su vez con (*condicion1*) **and/or** (*condicion2*), lo cual permite chequar condiciones más complejas:

```python
x = int(input("Ingrese el primer número:"))
y = int(input("Ingrese el segundo número:"))
z = ( x> 15 or x <-15) and ( y > 15 or y < -15 )
print("El valor absoluto de ambos numeros es mayor a 15?", z)
```

### 9.1. Operadores Módulo y Cociente Entero


El **operador módulo** permite calcular el resto en una división de un número por otro. Para aplicarlo se utiliza el símbolo **%**. Por ejemplo:

*   $5$ % $3$ $= 2$
*   $17$ % $10$ $= 7$
*   $9$ % $6$ $= 3$
*   $15$ % $5$ $= 0$

Una aplicación de utilidad consiste en preguntar si el resto de un numero **x** en la division por **y** es igual a 0 y, de esta forma, saber si **x** es múltiplo de **y**:

```python
x = 27
y = 9
print("Resultado:", x % y )
```

```python
x = int( input("Ingrese un número:") )
print("Es el número par?", x%2 == 0 )
print("Es el número multiplo de 3?", x%3 == 0 )
print("Es el número multiplo de 5?", x%5 == 0 )
print("Es el número multiplo de 7?", x%7 == 0 )
```

Notemos que para comparar una variable el símbolo que utilizamos fue el **`==`**. Este es un operador de **comparación**. Al comparar dos variables, obtenemos como resultado True o False, según el valor de verdad de la proposición.

Por otro lado, el operador **//** permite obtener el cociente truncado a cero decimales. Es decir, la parte entera del resultado de la división.

```python
x = 10
y = 3
print("Resultado:", x / y)
print("Resultado:", x // y )
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
import ipywidgets as widgets
from ipywidgets import interact

# Configuración de estilo visual
sns.set_theme(style="whitegrid")

# Asumiendo que 'data' ya fue cargado previamente desde Agro_DS.csv
# data = pd.read_csv('Agro_DS.csv')

@interact(
    rango_temp=widgets.FloatRangeSlider(
        value=[18.0, 30.0],
        min=float(np.floor(data['temperature'].min())),
        max=float(np.ceil(data['temperature'].max())),
        step=0.5,
        description='🌡️ Temp (°C):',
        disabled=False,
        continuous_update=False,  # Actualiza solo al soltar el slider
        orientation='horizontal',
        readout=True,
        readout_format='.1f',
        style={'description_width': 'initial'}
    ),
    rango_lluvia=widgets.FloatRangeSlider(
        value=[50.0, 180.0],
        min=float(np.floor(data['rainfall'].min())),
        max=float(np.ceil(data['rainfall'].max())),
        step=5.0,
        description='🌧️ Lluvia (mm):',
        disabled=False,
        continuous_update=False,
        orientation='horizontal',
        readout=True,
        readout_format='.0f',
        style={'description_width': 'initial'}
    )
)
def filtrar_cultivos_por_clima(rango_temp, rango_lluvia):
    temp_min, temp_max = rango_temp
    rain_min, rain_max = rango_lluvia

    # 1. Filtrar el DataFrame según los límites de los dos sliders
    df_filtrado = data[
        (data['temperature'] >= temp_min) & (data['temperature'] <= temp_max) &
        (data['rainfall'] >= rain_min) & (data['rainfall'] <= rain_max)
    ]

    # Conteo de muestras que caen en la región filtrada por cultivo
    conteo_cultivos = df_filtrado['label'].value_counts().reset_index()
    conteo_cultivos.columns = ['Cultivo', 'Muestras compatibles']

    # 2. Resumen informativo en consola
    print("=" * 60)
    print(f"🎯 FILTRADO CLIMÁTICO DINÁMICO")
    print("=" * 60)
    print(f"• Rango de Temperatura : {temp_min:.1f}°C  ➜  {temp_max:.1f}°C")
    print(f"• Rango de Lluvia       : {rain_min:.0f} mm  ➜  {rain_max:.0f} mm")
    print(f"• Registros viables     : {len(df_filtrado)} / {len(data)} muestras")
    print(f"• Diversidad de cultivos: {len(conteo_cultivos)} tipos encontrados")
    print("=" * 60)

    if len(conteo_cultivos) == 0:
        print("\n⚠️ No se encontraron cultivos en el dataset que soporten este rango extremo simultáneamente.")
        return

    # 3. Creación de gráficos
    fig, axes = plt.subplots(1, 2, figsize=(16, 6))

    # Gráfico A: Top cultivos con más muestras compatibles
    top_cultivos = conteo_cultivos.head(10)
    sns.barplot(
        data=top_cultivos,
        x='Muestras compatibles',
        y='Cultivo',
        palette='Greens_r',
        ax=axes[0]
    )
    axes[0].set_title('Top 10 Cultivos Viables para este Clima', fontsize=13, weight='bold')
    axes[0].set_xlabel('Cantidad de Muestras en Rango')
    axes[0].set_ylabel('Cultivo')

    # Gráfico B: Scatter Plot con la región delimitada por líneas punteadas
    # Fondo: todos los datos en gris tenue
    sns.scatterplot(
        data=data,
        x='temperature',
        y='rainfall',
        color='lightgray',
        alpha=0.4,
        s=30,
        label='Fuera del filtro',
        ax=axes[1]
    )
    # Frente: datos dentro del rango destacado por cultivo
    sns.scatterplot(
        data=df_filtrado,
        x='temperature',
        y='rainfall',
        hue='label',
        palette='tab10',
        alpha=0.9,
        s=50,
        ax=axes[1]
    )

    # Dibujar cuadro/líneas punteadas del rango delimitado
    axes[1].axvline(temp_min, color='#d95f02', linestyle='--', linewidth=1.5)
    axes[1].axvline(temp_max, color='#d95f02', linestyle='--', linewidth=1.5)
    axes[1].axhline(rain_min, color='#7570b3', linestyle='--', linewidth=1.5)
    axes[1].axhline(rain_max, color='#7570b3', linestyle='--', linewidth=1.5)

    axes[1].set_title('Delimitación del Rango en la Población de Datos', fontsize=13, weight='bold')
    axes[1].set_xlabel('Temperatura (°C)')
    axes[1].set_ylabel('Lluvia (mm)')
    axes[1].legend(bbox_to_anchor=(1.05, 1), loc='upper left', fontsize='small', frameon=True)

    plt.tight_layout()
    plt.show()
```
